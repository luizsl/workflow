```python
import matplotlib.pyplot as plt
import numpy as np
from astropy.io import fits
from astropy.wcs import WCS

file_path = '../data_products/NGC613/miles/ifscube_4/input_cube_4_linefit.fits'

hdul = fits.open(file_path)
hdul.info()

solution = hdul['solution'].data
#names = hdul['parnames'].data
```

```python
file_path = '../data/NGC613/Muse/NGC0613_DATACUBE_FINAL_clean.fits.gz'
with fits.open(file_path) as hdu:
    header = hdu[1].header
    
wcs = WCS(header)
```

```python
plt.style.use('../exploration/fig_conf.mplstyle')
velocity_ha = solution[1]

ax = plt.subplot(projection=wcs[0])
im = ax.imshow(velocity_ha, origin = 'lower', cmap = 'Spectral_r')

cbar = plt.colorbar(im)
cbar.set_label(r'km/s')

plt.savefig('velocity_ha.pdf')
plt.show()
```

```python
#%matplotlib widget

plt.style.use('../exploration/fig_conf.mplstyle')
amplitude_ha = np.arcsinh(solution[0])

ax = plt.subplot(projection=wcs[0])
im = ax.imshow(amplitude_ha, origin = 'lower', vmin = 4, vmax = 11)

cbar = plt.colorbar(im)
cbar.set_label(r'unidades arbitrarias')

plt.savefig('amplitude_ha.pdf')
plt.show()
```
