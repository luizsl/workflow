```python
import numpy as np
from astropy.io import fits
import matplotlib.pyplot as plt
```

```python
# Fit with complete configuration
ngc1097_metal_path_miles = ('../../data_products/NGC1097_DATACUBE_FINAL_clean/MilesAgeMh/ppxf/mean_mh_light.fits')

with fits.open(ngc1097_metal_path_miles) as hdul:
    print('\nmiles metallicity')
    hdul.info()
    metal_miles = hdul[0].data
    
    
ngc1097_age_path_miles = ('../../data_products/NGC1097_DATACUBE_FINAL_clean/MilesAgeMh/ppxf/mean_log10_age_light.fits')
with fits.open(ngc1097_age_path_miles) as hdul:
    print('\nmiles age')
    hdul.info()
    age_miles = hdul[0].data

ngc1097_kin_path_miles = ('../../data_products/NGC1097_DATACUBE_FINAL_clean/MilesAgeMh/ppxf/sol.fits')
with fits.open(ngc1097_kin_path_miles) as hdul:
    print('\nmiles kinematics')
    hdul.info()
    sol_miles = hdul[0].data
```

```python
# Fit removing reddening correction
ngc1097_metal_path_miles_rm_reddening = ('../../data_products/NGC1097_DATACUBE_FINAL_clean/MilesAgeMh_8/ppxf/mean_mh_light.fits')
with fits.open(ngc1097_metal_path_miles_rm_reddening) as hdul:
    print('\nmiles metallicity')
    hdul.info()
    metal_miles_rm_reddening = hdul[0].data
    
    
ngc1097_age_path_miles_rm_reddening = ('../../data_products/NGC1097_DATACUBE_FINAL_clean/MilesAgeMh_8/ppxf/mean_log10_age_light.fits')
with fits.open(ngc1097_age_path_miles_rm_reddening) as hdul:
    print('\nmiles age')
    hdul.info()
    age_miles_rm_reddening = hdul[0].data
    
ngc1097_kin_path_miles = ('../../data_products/NGC1097_DATACUBE_FINAL_clean/MilesAgeMh_8/ppxf/sol.fits')   
with fits.open(ngc1097_kin_path_miles) as hdul:
    print('\nmiles kinematics')
    hdul.info()
    sol_miles_rm_reddening = hdul[0].data
```

```python
# Fit removing reddening correction with MILES-BASTI

ngc1097_metal_path_miles_basti_rm_reddening = ('../../data_products/NGC1097_DATACUBE_FINAL_clean/MilesAgeMhAlpha_1/ppxf/average_metallicity.fits')
with fits.open(ngc1097_metal_path_miles_basti_rm_reddening) as hdul:
    print('\nmiles metallicity')
    hdul.info()
    metal_miles_basti_rm_reddening = hdul[0].data
    
ngc1097_age_path_miles_basti_rm_reddening = ('../../data_products/NGC1097_DATACUBE_FINAL_clean/MilesAgeMhAlpha_1/ppxf/average_age.fits')
with fits.open(ngc1097_age_path_miles_basti_rm_reddening) as hdul:
    print('\nmiles age')
    hdul.info()
    age_miles_basti_rm_reddening = hdul[0].data
    
ngc1097_kin_path_miles_basti =              ('../../data_products/NGC1097_DATACUBE_FINAL_clean/MilesAgeMhAlpha_1/ppxf/sol.fits')   
with fits.open(ngc1097_kin_path_miles_basti) as hdul:
    print('\nmiles kinematics')
    hdul.info()
    sol_miles_basti_rm_reddening = hdul[0].data
```

```python
# Fit removing dynamic masking
ngc1097_metal_path_miles_rm_dynmask = ('../../data_products/NGC1097_DATACUBE_FINAL_clean/MilesAgeMh_7/ppxf/mean_mh_light.fits')
with fits.open(ngc1097_metal_path_miles_rm_dynmask) as hdul:
    print('\nmiles metallicity')
    hdul.info()
    metal_miles_rm_dynmask = hdul[0].data
    
    
ngc1097_age_path_miles_rm_dynmask = ('../../data_products/NGC1097_DATACUBE_FINAL_clean/MilesAgeMh_7/ppxf/mean_log10_age_light.fits')
with fits.open(ngc1097_age_path_miles_rm_dynmask) as hdul:
    print('\nmiles age')
    hdul.info()
    age_miles_rm_dynmask = hdul[0].data
    
ngc1097_kin_path_miles_rm_dynmask = ('../../data_products/NGC1097_DATACUBE_FINAL_clean/MilesAgeMh_7/ppxf/sol.fits')   
with fits.open(ngc1097_kin_path_miles_rm_dynmask) as hdul:
    print('\nmiles kinematics')
    hdul.info()
    sol_miles_rm_dynmask = hdul[0].data
```

```python
%matplotlib widget

fig, ax = plt.subplots(2, 4, constrained_layout=True, figsize=(8,4), sharex=True, sharey=True)

# Age
im0=ax[0,0].imshow(10**(age_miles-9), origin='lower', vmin=2.5, vmax=12)
ax[0,1].imshow(10**(age_miles_rm_reddening-9), origin='lower', vmin=2.5, vmax=12)
ax[0,2].imshow(10**(age_miles_rm_dynmask-9), origin='lower', vmin=2.5, vmax=12)
ax[0,3].imshow(age_miles_basti_rm_reddening, origin='lower', vmin=2.5, vmax=12)

# Metallicity
im1=ax[1,0].imshow(metal_miles, origin='lower', vmin=-0.8, vmax=0.2)
ax[1,1].imshow(metal_miles_rm_reddening, origin='lower', vmin=-0.8, vmax=0.2)
ax[1,2].imshow(metal_miles_rm_dynmask, origin='lower', vmin=-0.8, vmax=0.2)
ax[1,3].imshow(metal_miles_basti_rm_reddening, origin='lower', vmin=-0.8, vmax=0.2)

ax[0,0].set_title('E-MILES-Padova \n')
ax[0,1].set_title('E-MILES-Padova \nw/o redding')
ax[0,2].set_title('E-MILES-Padova \nw/o dyn. mask.')
ax[0,3].set_title('MILES-BASTI \nw/o reddening')

ax[0,0].set_ylabel('<age>')
ax[1,0].set_ylabel('<[M/H]>')

fig.colorbar(im0, ax= ax[0,:])
fig.colorbar(im1, ax= ax[1,:])
```

```python
fig, ax = plt.subplots(4, 4, constrained_layout=True, figsize=(8,8), sharex=True, sharey=True)

# V
im2=ax[0,0].imshow(sol_miles[0], origin='lower', vmin=-180, vmax=180)
ax[0,1].imshow(sol_miles_rm_reddening[0], origin='lower', vmin=-180, vmax=180)
ax[0,2].imshow(sol_miles_rm_dynmask[0], origin='lower', vmin=-180, vmax=180)
ax[0,3].imshow(sol_miles_basti_rm_reddening[0], origin='lower', vmin=-180, vmax=180)

# sigma
im3=ax[1,0].imshow(sol_miles[1], origin='lower', vmin=80, vmax=250)
ax[1,1].imshow(sol_miles_rm_reddening[1], origin='lower', vmin=80, vmax=250)
ax[1,2].imshow(sol_miles_rm_dynmask[1], origin='lower', vmin=80, vmax=250)
ax[1,3].imshow(sol_miles_basti_rm_reddening[1], origin='lower', vmin=80, vmax=250)

# h3
im4=ax[2,0].imshow(sol_miles[2], origin='lower', vmin=-0.22, vmax=0.22)
ax[2,1].imshow(sol_miles_rm_reddening[2], origin='lower', vmin=-0.22, vmax=0.22)
ax[2,2].imshow(sol_miles_rm_dynmask[2], origin='lower', vmin=-0.22, vmax=0.22)
ax[2,3].imshow(sol_miles_basti_rm_reddening[2], origin='lower', vmin=-0.22, vmax=0.22)

# #h4
im5=ax[3,0].imshow(sol_miles[3], origin='lower', vmin=0.0, vmax=0.22)
ax[3,1].imshow(sol_miles_rm_reddening[3], origin='lower', vmin=0.0, vmax=0.22)
ax[3,2].imshow(sol_miles_rm_dynmask[3], origin='lower', vmin=0.0, vmax=0.22)
ax[3,3].imshow(sol_miles_basti_rm_reddening[3], origin='lower', vmin=0, vmax=0.22)
               
ax[0,0].set_title('E-MILES-Padova \n')
ax[0,1].set_title('E-MILES-Padova \nw/o redding')
ax[0,2].set_title('E-MILES-Padova \nw/o dyn. mask.')
ax[0,3].set_title('MILES-BASTI \nw/o reddening')

ax[0,0].set_ylabel(r'V$_{\star}$')
ax[1,0].set_ylabel(r'$\sigma_{\star}$')
ax[2,0].set_ylabel(r'h$_3$')
ax[3,0].set_ylabel(r'h$_4$')

fig.colorbar(im2, ax= ax[0,:])
fig.colorbar(im3, ax= ax[1,:])
fig.colorbar(im4, ax= ax[2,:])
fig.colorbar(im5, ax= ax[3,:])
```
