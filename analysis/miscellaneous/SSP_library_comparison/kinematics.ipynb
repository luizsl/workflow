```python
%matplotlib widget
# Reading data products of fit with XSL Padova

path = os.path.join(XSL_P00_path, kinematics_filepath)
with fits.open(path) as hdl:
    kinematics_xsl_p00 = hdl[0].data

# Reading data products of fit with XSL Parsec/Colibri    
path = os.path.join(XSL_PC_path, kinematics_filepath)
with fits.open(path) as hdl:
    kinematics_xsl_pc = hdl[0].data
    
# Reading data products of fit with XSL Parsec/Colibri    
path = os.path.join(MILES_BASTI_path, kinematics_filepath)
with fits.open(path) as hdl:
    kinematics_miles_basti = hdl[0].data
```

```python
fig, ax = plt.subplots(4, 3, figsize=(7, 6), constrained_layout=True, sharey=True, sharex=True)

for i, data in enumerate([kinematics_xsl_p00, kinematics_xsl_pc, kinematics_miles_basti]):
    with plt.style.context('science'):
        # Plot V_star
        kwargs = {'origin': 'lower', 'vmin': -110, 'vmax': 110, 'cmap': 'coolwarm'
                 }

        im = ax[0, i].imshow(data[0], **kwargs)
        cb = plt.colorbar(im, ax=ax[0, i])
        cb.set_label(r'V$_\star \, [\rm{kms}^{-1}]$',)
        # cb.ax.tick_params(rotation=90)
        ax[0, i].axes.get_yaxis().set_visible(False)
        ax[0, i].axes.get_xaxis().set_visible(False)

        # Plot sigma_star
        kwargs = {'origin': 'lower', 'vmin': 50,'vmax': 120,
                 }

        im = ax[1, i].imshow(data[1], **kwargs)
        cb = plt.colorbar(im, ax=ax[1, i])
        cb.set_label(r'$\rm{\sigma}_\star \, [\rm{kms}^{-1}]$')
        # cb.ax.tick_params(rotation=90)
        ax[1, i].axes.get_yaxis().set_visible(False)
        ax[1, i].axes.get_xaxis().set_visible(False)
        
        # Plot h_3
        kwargs = {'origin': 'lower', 'vmin': -0.15,'vmax': 0.15, 
                 }

        im = ax[2, i].imshow(data[2], **kwargs)
        cb = plt.colorbar(im, ax=ax[2, i])
        cb.set_label('h$_3$')
        # cb.ax.tick_params(rotation=90)
        ax[2, i].axes.get_yaxis().set_visible(False)
        ax[2, i].axes.get_xaxis().set_visible(False)
        
        # Plot h_4
        kwargs = {'origin': 'lower', 'vmin': -0.03,'vmax': 0.15,
                 }

        im = ax[3, i].imshow(data[3], **kwargs)
        cb = plt.colorbar(im, ax=ax[3, i])
        cb.set_label('h$_4$')
        # cb.ax.tick_params(rotation=90)
        ax[3, i].axes.get_yaxis().set_visible(False)
        ax[3, i].axes.get_xaxis().set_visible(False)

    ax[0, 0].set_title('XSL Padova\nKroupa uni.')
    ax[0, 1].set_title('XSL P/C\nKroupa uni.')
    ax[0, 2].set_title('Miles BaSTI\nKroupa rev.')
                                   
```
